# ❄️ SnowIO: Interactive Landsat Scene Query

This notebook provides an interactive interface for querying Landsat scenes based on:
- **User-drawn polygon** (max 100 vertices)
- **Date range** selection
- **Cloud cover** filtering

## Features
- 🗺️ Interactive map centered on Alaska
- 🛰️ Multiple basemap layers (OSM, RGB imagery, shaded relief)
- ✏️ Drawing tools for defining area of interest
- 📅 Date range selection
- 🔍 Real-time Landsat scene search

## Alternative: Manual Polygon Input

If you prefer to manually specify polygon coordinates, you can use the method below.
Coordinates should be provided as [longitude, latitude] pairs.

```python
# Example: Set a polygon around Anchorage, Alaska
# Format: [[lon, lat], [lon, lat], ...]
anchorage_polygon = [
    [-150.5, 61.5],
    [-149.0, 61.5],
    [-149.0, 60.8],
    [-150.5, 60.8],
    [-150.5, 61.5]  # Close the polygon
]

# Set the polygon
query_map.set_polygon(anchorage_polygon)
print("✅ Polygon set successfully!")
print(f"   Vertices: {len(anchorage_polygon)}")
```

## View Results Summary

Access the results programmatically for custom analysis.

```python
# Display summary statistics
if query_map.results:
    print(f"Total scenes found: {len(query_map.results)}")
    print(f"\nFirst scene details:")
    print(f"  Scene ID: {query_map.results[0].get('display_id', 'N/A')}")
    print(f"  Date: {query_map.results[0].get('acquisition_date', 'N/A')}")
    print(f"  Cloud Cover: {query_map.results[0].get('cloud_cover', 'N/A')}%")
    
    # Show date distribution
    dates = [s.get('acquisition_date') for s in query_map.results if s.get('acquisition_date')]
    if dates:
        print(f"\nDate range in results: {min(dates)} to {max(dates)}")
else:
    print("No results yet. Run a query first!")
```

## Add Results to Map

Visualize the scene locations on the map.

```python
# Create a new map with results
if query_map.results:
    results_map = query_map.create_map()
    query_map.add_results_to_map()
    print(f"✅ Created map with {len(query_map.results)} scene locations!")
    display(results_map)
else:
    print("No results to display. Run a query first!")
```

## Tips and Notes

### Drawing Tools
- **Polygon Tool**: Click multiple points to create a custom polygon
- **Rectangle Tool**: Click and drag to create a rectangular area
- **Edit Tool**: Modify existing polygons by dragging vertices
- **Delete Tool**: Remove drawn polygons
- **Export Button**: Save your polygon as GeoJSON (top right corner)

### Map Layers
- **OpenStreetMap**: Best for identifying place names, roads, and boundaries
- **ESRI World Imagery**: High-resolution satellite imagery (RGB)
- **Shaded Relief**: Topographic visualization showing elevation

### Query Parameters
- **Date Range**: Landsat 8 launched in 2013, Landsat 9 in 2021
- **Cloud Cover**: Lower values give clearer imagery but fewer results
- **Dataset**: Level 2 products are atmospherically corrected

### Polygon Limits
- Maximum 100 vertices per polygon
- Larger areas may return more scenes
- Consider cloud cover when selecting areas

### API Rate Limits
- USGS EarthExplorer API has rate limits
- Maximum 100 scenes per query
- Consider narrowing date range or area for better results

### Workflow Summary
1. **Draw**: Use the map drawing tools to define your area
2. **Export**: Click Export button, save as `polygon.geojson`
3. **Load**: Run the load cell to import your polygon
4. **Query**: Use widgets or direct API call to search scenes
5. **Export**: Save results as JSON for further analysis
